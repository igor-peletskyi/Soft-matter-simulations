```python

data_path = '/content/Softmatter_simulation_Introductory_course/practical_session_1/data'
import numpy as np
import matplotlib.pyplot as plt
import os
distance_file = os.path.join(data_path, 'distance_data_headers.csv')

distances = np.genfromtxt(
    fname=distance_file,
    delimiter=',',
    dtype=str
)
headers = distances[0]
data = distances[1:].astype(float)
print("Column names:", headers)
print("Data dimensions:", data.shape)
print("First five rows:")
print(data[:5])

sample = headers[1]

for col in range(1, len(headers)):
    fig = plt.plot(data[:,col], label=headers[col])
    plt.legend()

plt.xlabel('Simulation Frame')
plt.ylabel('Distance (angstrom)')
plt.savefig('all_samples.png', dpi=300)
data_min = np.min(data[:,1:])*0.9
data_max = np.max(data[:,1:])*1.1
for col in range(1, len(headers)):
   plt.figure()
   sample = headers[col]

   plt.plot(data[:,col], '--', label=sample)
   plt.ylim(data_min, data_max)
   plt.legend()
   plt.xlabel('Simulation Frame')
   plt.ylabel('Distance (angstrom)')
   plt.savefig(F'{sample}.png')


```
